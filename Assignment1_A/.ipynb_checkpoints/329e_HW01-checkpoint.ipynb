## Q1 - How many unique Airlines are there in this dataset? (2 points)

## Q2 - What is the mean departure delay of all flights considering only the positive departure delay values? (2 points)
Negative delay value means that the flight left earlier than scheduled. 




```python
# Add your code here! 
positive_delays = df[df['DEPARTURE_DELAY'] > 0]['DEPARTURE_DELAY']
num = positive_delays.mean()

print('The mean delay is of', num, 'minutes')
```

# Q4 - Find out which origin airport has the largest departure delay on Weekends. (2 points)
Weekend days have the IDs of DAY_OF_WEEK 6 and 7. 


```python
# filter by weekends
weekends = df[(df['DAY_OF_WEEK'] == 6) | (df['DAY_OF_WEEK'] == 7)]

# find max
largest_delay = weekends['DEPARTURE_DELAY'].idxmax()

# identify which airport has max
print('Airport with largest single departure delay:', weekends.loc[largest_delay, 'ORIGIN_AIRPORT'])
```

# Q5 - Find the top 3 Airlines which had the largest mean departure delay in the month of January (2 points)
Month of January has the id of 1. Consider only flights with the positive departure delay. 

```python
# filter month
january = df[df['MONTH'] == 1]

# filter delay
delayed = january[january['DEPARTURE_DELAY'] > 0]

# group by mean
mean_delays = delayed.groupby('AIRLINE')['DEPARTURE_DELAY'].mean()

# Sort from largest to smallest and take the top 3
top_3 = mean_delays.sort_values(ascending=False).head(3)

print('Airlines with the largest mean departure delay are: \n',top_3)
```

## Q6 - What percentage of flights out of Austin (AUS) are canceled? (2 points)
The column field "CANCELLED" will be equal to one of the flight is cancelled. 

```python
#filter to origin airports from AUS
austinFlights = df[df['ORIGIN_AIRPORT'] == 'AUS']

# find number of cancelled
cancelled = austinFlights[austinFlights['CANCELLED'] == 1]

# compare to total from AUS
percentage = len(cancelled) / len(austinFlights) 
percentage
print(f"{percentage:.5f}% of flights out of AUS are cancelled")

```

# Q7 - Which origin airport has the most cancellation of departure flights? (2 points)


```python
# filter by cancellations
cancellations = df[df['CANCELLED'] == 1]

# group cancellations by origin airport
cancByOrg = cancellations.groupby('ORIGIN_AIRPORT').size()

# find max cancellations
mostCanc = cancByOrg.idxmax()
mostCanc
print(f"{mostCanc} has the most cancellation of departure flights")
```

## Q8 - Create a histogram plot that shows the data distribution of the arrival delays consider only positive arrival delays and smaller than 200 minutes. Use 10 minutes bin/bucket size. (2 points)

```python
# filter to delays between 0 and 200
arr_delay = df['ARRIVAL_DELAY'].dropna()
filtered_delay = arr_delay[(arr_delay > 0) & (arr_delay < 200)]

# bins of size 10
bins = np.arange(0, 210, 10)

plt.hist(filtered_delay, bins=bins)
plt.xlabel('Arrival Delay (minutes)')
plt.ylabel('Number of Flights')
plt.title('Distribution of Positive Arrival delays (< 200 min)')
plt.xticks(bins)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Q9 -  Plot a scatter plot of  flight travel air time and distance. Describe if there is any corrolation between them. (2 points)

Describe and interpret the correlation between the two variables of distance and air travel time. 

```python
# filter to both air time and distance then grab both
flight_data = df[['DISTANCE', 'AIR_TIME']].dropna()
distance = flight_data['DISTANCE']
air_time = flight_data['AIR_TIME']

# super awesome correlation function
correlation = np.corrcoef(distance, air_time)[0, 1]
# distance on x, air time on y
plt.scatter(distance, air_time, alpha=0.3)
plt.xlabel('Distance (miles)')
plt.ylabel('Air Time (minutes)')
plt.title('Relationship Between Flight Distance and Air Time')
plt.grid(alpha=0.1)
plt.tight_layout()
plt.show()
```

```python
print('The correlation coefficient is ', correlation, 'indicating high correlation between flight distance and air time. As the distance in miles increases, so too will the air time.')
```

# Question - 10: Which top-3 destination cities has the highst average departure delays on Weekends? (2 points)

* Consider only the positive departure delays. 
* Weekend flights have a day of week of 6 and 7
* You need to merge the airport dataset with the flight dataset to get the destination city of each flight
* You need to group by city and calculate the mean departure delays 
* At the end print out the top 3 of the departure delays with the average delays

```python
# Add your code here!
# filter to weekends
weekend_flights = df[df["DAY_OF_WEEK"].isin([6, 7])]

# only positive departure delays
weekend_flights = weekend_flights[weekend_flights["DEPARTURE_DELAY"] > 0]

# merge with airports
merged = pd.merge(weekend_flights, airports, left_on="DESTINATION_AIRPORT", right_on="iata", how="inner")

#groupby city, sort by delay
city_avg_delay = merged.groupby("city")['DEPARTURE_DELAY'].mean()
top3 = city_avg_delay.sort_values(ascending=False).head(3)

print("Top 3 destination cities with highest average departure delay on weekends:")
print(top3.to_string())
```
