```python
import numpy as np
from scipy.stats import t
import scipy
from scipy import stats
```

```python
help(stats.ttest_1samp)
```

```python
data = [14.2, 15.1, 13.8, 14.5, 14.8, 15.4, 13.9, 14.6, 14.1, 14.7]
data
```

STEP 1

H0 : mu = 15 (battery life is the same as it's claimed)

H1 : mu != 15 (the battery life is different than it's claimed)

STEP 2

alpha = 0.05

STEP 3

Calculate test statistic

```python
s_mean = np.mean(data)
s_std = np.std(data, ddof = 1)
se = s_std / np.sqrt(10)
t_stat = (s_mean - 15) / se
s_mean,s_std,se, t_stat
```

STEP 4

Find crit and p-values

```python
t_crit = stats.t.ppf(1 - 0.05 / 2, 9)
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), 9))
t_crit, p_value
```

```python
stats.ttest_1samp(data,15,alternative = 'two-sided')
```

CONCLUSION:

Because the p-value(0.015) is smaller than the alpha(0.05), we can reject the null hypothesis. The t-value is negative, meaning there is significant data to suggest the battery life is less than 15 hours long.
