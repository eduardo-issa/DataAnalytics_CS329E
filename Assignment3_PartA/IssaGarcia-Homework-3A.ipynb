```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

***(1)***

```python
df = pd.read_csv("data.csv")
days = df["days"]
print(days)
```

***(2)***

```python
bin_edges = np.arange(days.min(), days.max() + 2, 1)
fig, ax = plt.subplots()
ax.hist(days, bins=bin_edges, edgecolor="black", align="left")
ax.set_title("Distribution of Hospital Length of Stay")
ax.set_xlabel("Length of Stay in days")
ax.set_ylabel("Number of Patients")
plt.show()
```

The data shows a unimodal distribution with a right skew. The majority of patients spend 3-6 days in the hospital, and the center hovers around 4-5 days.

***(3)***

```python
mean = days.mean()
median = days.median()
std = days.std()

mean,median,std
```

```python
q1, q3 = days.quantile([0.25, 0.75])
iqr = q3 - q1

q1,q3,iqr, days.min(), days.max()
```

On average, a patient spends 5.45 days in hospital and the highest number of patients stay for 5 days. The mean is pulled slighlty above the median due to the long right tail, so median is the more appropriate center of the distribution.

a)


```python
mu = 5
sigma = 3

prob_a = stats.norm.cdf(7, loc=mu, scale=sigma)
print(f"Percentage of patients in the hospital for less than 7 days: {prob_a * 100:.2f}%")
```

b)

```python
n = 10
se = sigma / np.sqrt(n)

prob_b = 1 - stats.norm.cdf(7, loc=mu, scale=se)
print(f"Probability of observing sample mean greater than 7: {prob_b * 100:.2f}%, assuming data is consistent with the truth.")
```
