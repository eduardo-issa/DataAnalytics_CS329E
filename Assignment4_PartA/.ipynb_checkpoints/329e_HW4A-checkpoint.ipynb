## Task - 5  Correlation Test $\rho$-Test 
Do a Hypothesis test to determine whether there  is a statistically significant correlation between Heart Rate and Temperature. Provide your conclusion and your p-value for the test.  **(4 points)**

```python
# Add your Code Here 
from scipy import stats

# alpha = 0.05

# H0 rho = 0 (there is no linear association)
# H1 rho != 0 (there is a linear association)

r_stat, p_val = stats.pearsonr(df['Temperature'], df['HeartRate']) 
n = len(df) 
df_deg = n - 2 
alpha = 0.05

t_stat = r_stat * np.sqrt(df_deg / (1 - r_stat**2)) 
t_crit = stats.t.ppf(1 - alpha / 2, df_deg)

print(f'Pearson r: {r_stat:.4f}')
print(f't-statistic: {t_stat:.4f} (Critical t: ±{t_crit:.4f})') 
print(f'p-value: {p_val:.5f}')
```

At the alpha = 0.05 significance level, we reject H0 (t = 2.967, p = 0.0036$). There is statistically significant evidence of a positive linear correlation between Body Temperature and Heart Rate.
