```python
import numpy as np
import pandas as pd
from scipy.stats import t
import scipy
from scipy import stats
import matplotlib as plt
```

```python
df = pd.read_csv("ICA4.csv")
df
```

```python
group_a = df[df["Study_Method"] == "Traditional"]["Homework_Time_Minutes"]
group_b = df[df["Study_Method"] == "Study_App"]["Homework_Time_Minutes"]
```

```python
nT, nS = len(group_a), len(group_b)
meanT, meanS = np.mean(group_a), np.mean(group_b)
stdT, stdS = np.std(group_a, ddof=1), np.std(group_b, ddof=1)
se = np.sqrt(stdT**2/nT + stdS**2/nS)

meanT, meanS, stdT, stdS, se
```

```python
t_stat = (meanT - meanS) / se
t_stat
```

```python
t_crit = stats.t.ppf(1 - 0.05/2, 24)
t_crit
```

```python
stats.ttest_ind(group_a, group_b)
```

We can reject the null hypothesis because the p-value is larger than the alpha (0.05). The t-stat is positive, so evidence suggests the study app works better. The assumption that the students were put into samples randomly holds, as well as the assumption that both are normal distributions.
